```python
import matplotlib.pyplot as plt

df['quality'].value_counts().sort_index().plot(
    kind='bar', edgecolor='black'
)

plt.title('Wine Quality')
plt.xlabel('Quality')
plt.ylabel('Number of Wines')
plt.show()
```

```python
plt.scatter(df['alcohol'], df['quality'])

plt.title('Alcohol vs Wine Quality')
plt.xlabel('Alcohol')
plt.ylabel('Quality')
plt.show()


```

```python
quality_count = df['quality'].value_counts().sort_index()

plt.plot(quality_count.index, quality_count.values, marker='o')

plt.title('Wine Quality Trend')
plt.xlabel('Quality')
plt.ylabel('Number of Wines')
plt.grid(True)
plt.show()
```
